### 1. Scaling: three different approaches

Compare Min-Max, Standard and Robust scaling.

Let's first recap:

#### Min–Max scaling

$$
x_{\text{scaled}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}
$$

Subtract the minimum and divide by the range. Values in the data used to fit the scaler fall between **0 and 1**.

#### Standard scaling (Z-score)

$$
x_{\text{scaled}} = \frac{x - \bar{x}}{s}
$$

Here, $\bar{x}$ is the mean and $s$ is the standard deviation. The scaled variable has **mean 0 and standard deviation 1**.

#### Robust scaling

$$
x_{\text{scaled}} =
\frac{x - \operatorname{median}(x)}{\operatorname{IQR}(x)}
$$

Here, $\operatorname{IQR}(x) = Q_{75} - Q_{25}$ is the interquartile range: the difference between the 75th and 25th percentiles. The scaled variable has **median 0 and IQR 1**. This method is less sensitive to extreme values.

```python
# Packages for this exercise. If one is missing, install it once from a terminal
# (in a Jupyter notebook, type %pip instead of pip):
# pip install pandas numpy matplotlib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

hospital = pd.read_csv("data/hospital_admissions.csv")
```

### From a calculation to a function

To apply Min-Max scaling to CRP, we can write the calculation directly:

```python
crp = hospital["CRP_mg_L"]

hospital.assign(
    CRP_scaled=(crp - crp.min()) / (crp.max() - crp.min())
).head()
```

To reuse this calculation, we can give it a name by creating a **function**:

```python
def minmax(x):
    return (x - x.min()) / (x.max() - x.min())
```

A function **takes an input, performs a calculation and returns the result**.
Here, `x` is a placeholder for the numerical column we want to scale.
`def` starts the function, the indented lines below it are its body, and
`return` hands back the result. Unlike R, Python needs the word `return`.

We can now call our function inside `assign()`, the pandas counterpart of
`mutate()`:

```python
hospital.assign(
    CRP_scaled=minmax(hospital["CRP_mg_L"]),
    age_scaled=minmax(hospital["age"])
).head()
```

Each column takes the place of `x`. We write the calculation once and
reuse it for different variables.

Let's build the remaining two functions of the three scaling methods and apply them to CRP.

```python
def standard(x):
    return (x - x.mean()) / x.std()

def robust(x):
    return (x - x.median()) / (x.quantile(0.75) - x.quantile(0.25))
```

pandas ignores missing values when it calculates the minimum, mean and
other summary statistics, which is what `na.rm = TRUE` does in R.

### Apply each function to CRP

We pass the same CRP column to each function and save the results as
three new columns:

```python
hospital_scaled = hospital.assign(
    CRP_minmax=minmax(hospital["CRP_mg_L"]),
    CRP_standard=standard(hospital["CRP_mg_L"]),
    CRP_robust=robust(hospital["CRP_mg_L"])
)
```

```python
print(hospital_scaled.filter(like="CRP").describe().round(3))

print(hospital_scaled["CRP_standard"].std())
```

Visualise CRP before and after scaling:

```python
cols = ["CRP_mg_L", "CRP_minmax", "CRP_standard", "CRP_robust"]

hospital_scaled[cols].hist(bins=30, figsize=(9, 6))
plt.tight_layout()
plt.show()
```

> **Think about it**
>
> 1.  What happens to the range after Min-Max scaling?
> 2.  What happens to the mean and standard deviation after Standard scaling?
> 3.  How does Robust scaling differ?

> **Worked answer**
>
> **Min-Max** puts every value between 0 and 1: the lowest CRP, 0.8, becomes 0 and the highest, 264.6, becomes 1. Look at where the typical patient lands, though — the median is 0.037. Most of the cohort is crammed into the bottom tenth of the scale.
>
> **Standard** gives a mean of 0 and a standard deviation of 1. It is not bounded: the highest CRP sits about 8.4 standard deviations above the mean.
>
> **Robust** centres on the median instead of the mean, so the median becomes 0, and divides by the interquartile range instead of the standard deviation. The highest patient is 13.7 interquartile ranges out.

> **Key observation**
>
> All three methods change the numerical scale, but preserve the underlying shape of the distribution.

### 2. Outliers: what happens to the other patients?

Imagine a patient has an exceptionally high CRP. How does that influence the scaling of everyone else's measurements?

Compare the largest CRP values before and after Min-Max scaling:

```python
print(hospital_scaled["CRP_mg_L"].nlargest(5).tolist())

print(hospital_scaled["CRP_minmax"].nlargest(5).round(3).tolist())
```

Now make the maximum CRP three times larger, change nothing else, and scale again:

```python
crp = hospital["CRP_mg_L"]

hospital_extreme = hospital.assign(
    CRP_extreme=np.where(crp == crp.max(), crp * 3, crp)
)
hospital_extreme["CRP_minmax"] = minmax(hospital_extreme["CRP_extreme"]).round(3)

(hospital_extreme[["CRP_extreme", "CRP_minmax"]]
 .sort_values("CRP_extreme", ascending=False)
 .head(5))
```

> **Think about it**
>
> 1.  What happened to the scaled values of the other patients?
> 2.  Which scaling method would be least affected by an extreme CRP measurement?
> 3.  Should a clinically plausible extreme value automatically be removed?

> **Worked answer**
>
> **They all moved, though none of them changed.** The patient with the second-highest CRP, 197.7, was at 0.746 and is now at 0.248. Their measurement is exactly what it was. Min-Max divides by the range, the range is set by the single largest value, and so one patient decides where everybody else sits.
>
> **Robust**, and you can check it:

```python
hospital_extreme["robust_original"] = robust(hospital_extreme["CRP_mg_L"]).round(2)
hospital_extreme["robust_extreme"] = robust(hospital_extreme["CRP_extreme"]).round(2)

(hospital_extreme[["CRP_mg_L", "CRP_extreme", "robust_original", "robust_extreme"]]
 .sort_values("CRP_mg_L", ascending=False)
 .head(5))
```

> Only the first number is different. The second-highest patient is at 10.12 before and 10.12 after, because the median and the interquartile range are not moved by one extreme value. Standard scaling falls in between: the same patient drops from 6.1 to 3.2, since the standard deviation is inflated by the outlier.
>
> **No.** A CRP of 265 is not an error; in a real patient it would mean a severe infection, and probably the patient a model most needs to get right. Removing a value because it is inconvenient for a scaler throws away the cases that matter. Pick a scaler, or a transformation, that copes with them instead.

> **Key observation**
>
> Min-Max scaling is sensitive to extreme values. Robust scaling uses the median and interquartile range, which are far less sensitive to them.

### 3. Transformation: can we reduce skewness?

Scaling does not change the shape of a distribution. A nonlinear transformation can. Compare the original CRP distribution with its log transformation:

```python
hospital["log_CRP"] = np.log1p(hospital["CRP_mg_L"])

hospital[["CRP_mg_L", "log_CRP"]].hist(bins=30, figsize=(8, 3))
plt.tight_layout()
plt.show()
```

> **Think about it**
>
> 1.  How does the shape change?
> 2.  What happens to the distance between low and high CRP measurements?
> 3.  Why use `log1p()` rather than `log()`?

> **Worked answer**
>
> **The long right tail is pulled in** and the distribution becomes close to symmetric:
> CRP's skewness falls from 4.43 to 0.38.
>
> **Distances stop being about units and start being about ratios.** On the original scale, going from 5 to 10 mg/L is a step of 5, and going from 100 to 200 is a step of 100. After the transformation those two doublings are almost the same size: about 0.61 and 0.69. That matches how CRP is read clinically — a doubling means something whether it starts low or high — and it stops the handful of very high values from dominating everything measured in distance.
>
> **Because the log of zero is minus infinity.** `log1p(x)` computes $\log(1 + x)$, so a value of 0 becomes 0 instead of breaking the column. In this file the lowest CRP is 0.8, so a plain `log()` would have run — but a laboratory can report a result below its detection limit as 0, and `log1p()` is the habit that survives that.

### 4. Feature engineering: calculate BMI

Our dataset includes patients' height and weight, but not their Body Mass Index (BMI). BMI is widely used in clinical practice and medical research, as it provides a measure of body size relative to height and has been associated with a range of health outcomes. By combining height and weight into a single, clinically meaningful feature, we may capture information that helps our predictive model. This is an example of feature engineering, where existing variables are used to create potentially more informative predictors.

Let's calculate BMI and explore its distribution.

```python
hospital["BMI"] = hospital["weight_kg"] / (hospital["height_cm"] / 100) ** 2

hospital["BMI"].hist(bins=25, figsize=(7, 2.8))
plt.xlabel("BMI (kg/m²)")
plt.tight_layout()
plt.show()
```

> **Think about it**
>
> 1.  Why might BMI be more informative than weight alone?
> 2.  Is BMI adding new measurements, or combining existing information?
> 3.  What other clinically meaningful features could you derive from the variables you already have?

> **Worked answer**
>
> **Because weight means nothing without height.** Two patients from this file:

```python
(hospital[hospital["patient_id"].isin(["H134", "H184"])]
 [["patient_id", "height_cm", "weight_kg", "BMI"]]
 .round({"BMI": 1}))
```

> Both weigh about 79 kg. One is 142 cm tall and has a BMI of 39.0; the other is 201 cm and has a BMI of 19.6. On weight alone a model sees two identical patients. On BMI it sees one who is obese and one at the lean end of normal.
>
> **Combining.** No new measurement was taken; BMI is arithmetic on two columns that were already there. What it adds is the *relationship* between them, in the form a clinician would use. A model handed height and weight separately would have to discover that ratio for itself, and a linear model cannot.
>
> **Others from this table:** the shock index, heart rate divided by systolic blood pressure, which is used at the bedside to flag circulatory compromise; a flag for tachycardia (heart rate over 100) or hypotension (systolic below 90); age bands; the log of CRP you have just made. The test is the same each time: does the new column say something clinically meaningful that the raw columns only imply?

### What to take from this

-   **Scaling changes the numerical range of a variable, not its shape.** CRP was as skewed after all three scalers as before.
-   **Robust scaling is less sensitive to extreme values.** One inflated measurement moved every other patient under Min-Max and none under Robust.
-   **Transformation can reduce skewness and changes the relationships between values.** A log turned steps of units into steps of ratios.
-   **Feature engineering creates new variables from existing information.** BMI added no measurement and still separated two patients that weight could not.

> **Remember**
>
> Every minimum, maximum, mean, standard deviation, median and interquartile range in this exercise was computed from all 250 patients. When you build a predictive model, anything learned from the data — including a scaler — must be fitted on the training set only, not the full dataset.
